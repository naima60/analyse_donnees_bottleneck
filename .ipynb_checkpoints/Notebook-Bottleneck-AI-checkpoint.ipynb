```python
#===============================================================
# Fonction standard de téléchargement des fichiers
#=================================================================

def charger_fichier(chemin_fichier):
    """
    Charge un fichier à partir de son chemin complet, 
    affiche ses dimensions et retourne le DataFrame.
    """
    if not os.path.exists(chemin_fichier):
        raise FileNotFoundError(f"Fichier introuvable : {chemin_fichier}")
        
    df = pd.read_excel(chemin_fichier, engine="openpyxl")
    
    print(f"✓ Chargé avec succès : {chemin_fichier}")
    print(f"   • Dimensions (Shape) : {df.shape[0]} lignes | {df.shape[1]} colonnes")
    print("-" * 50)
    
    return df
```

<div style="border: 1px solid RGB(51,165,182);" >
<h3 style="margin: auto; padding: 20px; color: RGB(51,165,182); ">1.3 - Lecture des fichiers</h3>

```python
# ==========================================
# FONCTION D'INSPECTION UNIVERSELLE
# ==========================================

def inspecter_dataframe(df):
    """
    Affiche un rapport d'inspection standardisé pour n'importe quel DataFrame,
    sans nécessiter de paramètre de colonne :
    1. Résumé structurel (.info())
    2. Aperçu des 5 premières lignes (sans troncature)
    3. Vérification des lignes entièrement dupliquées
    
    Paramètre :
        df (pd.DataFrame) : Le DataFrame à inspecter.
    """
    print("=" * 60)
    print(" RAPPORT D'INSPECTION DU DATASET")
    print("=" * 60)
    
    # 1. Info du DataFrame
    print("\n[1] Résumé structurel (.info()) :")
    df.info()
    
    # 2. Affichage des 5 premières lignes
    print("\n[2] Aperçu des 5 premières lignes :")
    with pd.option_context('display.max_rows', None, 'display.max_columns', None):
        display(df.head(5))
        
    # 3. Vérification des doublons globaux (lignes identiques)
    nb_doublons = df.duplicated().sum()
    print(f"\n[3] Intégrité globale :")
    print(f"   • Nombre de lignes entièrement dupliquées : {nb_doublons}")
    
    print("=" * 60 + "\n")
```

<div style="border: 1px solid RGB(51,165,182);" >
<h3 style="margin: auto; padding: 20px; color: RGB(51,165,182); ">1.4 - Nettoyage des fichiers</h3>
</div>

```python
# ====================================
# FONCTION DE NETTOYAGE STANDARDISÉE
# ====================================

def nettoyer_dataframe(df):
    """
    Applique un nettoyage standardisé sur un DataFrame déjà chargé :
    1. Suppression des doublons
    2. Suppression des lignes entièrement vides
    3. Suppression des colonnes entièrement vides (NaN)
    4. Suppression des colonnes contenant uniquement des zéros (0)
    5. Suppression des colonnes constantes (<= 1 valeur unique)
    6. Séparation des colonnes datetime en Date et Heure distinctes
    
    Paramètre :
        df (pd.DataFrame) : Le DataFrame à nettoyer.
        
    Retourne :
        pd.DataFrame : Le DataFrame nettoyé.
    """
    print("=" * 60)
    print(" EXÉCUTION DU PIPELINE DE NETTOYAGE STANDARD")
    print("=" * 60)
    print(f"Dimensions avant nettoyage : {df.shape[0]} lignes | {df.shape[1]} colonnes")
    
    # Copie pour préserver le DataFrame d'origine
    df_clean = df.copy()
    
    # 1. Suppression des doublons
    nb_doublons = df_clean.duplicated().sum()
    df_clean = df_clean.drop_duplicates()
    print(f"✓ Doublons supprimés : {nb_doublons}")
    
    # 2. Suppression des lignes entièrement vides
    lignes_vides = df_clean.isna().all(axis=1).sum()
    df_clean = df_clean.dropna(how='all')
    print(f"✓ Lignes entièrement vides supprimées : {lignes_vides}")
    
    # 3. Suppression des colonnes entièrement vides (NaN)
    cols_vides = df_clean.columns[df_clean.isna().all()].tolist()
    df_clean = df_clean.dropna(axis=1, how='all')
    print(f"✓ Colonnes vides (NaN) supprimées : {cols_vides}")
    
    # 4. Suppression des colonnes contenant uniquement des zéros (0)
    cols_zeros = []
    for col in df_clean.columns:
        if pd.api.types.is_numeric_dtype(df_clean[col]) and (df_clean[col] == 0).all():
            cols_zeros.append(col)
    df_clean = df_clean.drop(columns=cols_zeros)
    print(f"✓ Colonnes remplies uniquement de 0 supprimées : {cols_zeros}")
    
    # 5. Suppression des colonnes constantes (une seule valeur unique)
    cols_constantes = [col for col in df_clean.columns if df_clean[col].nunique() <= 1]
    df_clean = df_clean.drop(columns=cols_constantes)
    print(f"✓ Colonnes constantes supprimées : {cols_constantes}")
    
    # 6. Séparation des datetime en Date et Heure
    datetime_cols = [col for col in df_clean.columns if pd.api.types.is_datetime64_any_dtype(df_clean[col])]
    for col in datetime_cols:
        df_clean[col + '_date'] = pd.to_datetime(df_clean[col]).dt.date
        df_clean[col + '_time'] = pd.to_datetime(df_clean[col]).dt.time
        df_clean = df_clean.drop(columns=[col])
        
    if datetime_cols:
        print(f"✓ Colonnes datetime séparées (Date / Heure) : {datetime_cols}")
        
    print(f"Dimensions finales après nettoyage : {df_clean.shape[0]} lignes | {df_clean.shape[1]} colonnes")
    print("-" * 60 + "\n")
    
    return df_clean
    
```

```python
# Téléchargement du fchier ERP
df_erp = charger_fichier("Sources/erp.xlsx")

# Infos fichier
print(inspecter_dataframe(df_erp))
```

```python
# ===================================
# VERIFICATION DU STATUT DE STOCK
# ==================================

# Valeurs uniques de la colonne stock_status
print(f"- Les valeurs uniques de la variable stack_status : \n {df_erp['stock_status'].unique()}")

# 1. Détermination du statut théorique basé sur la quantité en stock
# Règle métier : Si stock_quantity > 0 ==> instock, sinon outofstock
statut_theorique = np.where(df_erp['stock_quantity'] > 0, 'instock', 'outofstock')

# 2. Détection des incohérences de départ
masque_incoherence = df_erp['stock_status'] != statut_theorique
nb_incoherences_depart = masque_incoherence.sum()
nb_conformes_depart = len(df_erp) - nb_incoherences_depart

print(f"=== SITUATION DE DÉPART METIER ===")
print(f"- Nombre de lignes conformes stock_status au départ (True)  : {nb_conformes_depart} / {len(df_erp)}")
print(f"- Nombre de lignes INCOHÉRENTES stock_status au départ (False) : {nb_incoherences_depart} / {len(df_erp)}")

# 3. Affichage des lignes avec un statut incohérent au départ
print(f"\nDétail des {nb_incoherences_depart} anomalies détectées au départ :")
df_anomalies = df_erp[masque_incoherence].copy()
display(df_anomalies[['product_id', 'stock_quantity', 'stock_status']])

# 4. Correction directe du statut erroné
df_erp['stock_status'] = statut_theorique


# 5. Vérification post-correction (Contrôle programmatique)
anomalies_restantes = (df_erp['stock_status'] != np.where(df_erp['stock_quantity'] > 0, 'instock', 'outofstock')).sum()
assert anomalies_restantes == 0, "Erreur : Des incohérences subsistent après correction !"

print(f"\n=== BILAN POST-CORRECTION ===")
print(f"Nombre d'anomalies restantes : {anomalies_restantes}")
print("Le statut de stock du fichier ERP est désormais 100% cohérent avec les quantités réelles.")
```

```python
# ==========================================
#  TRAITEMENT METIER SPECIFIQUE ERP nettoyé
# ==========================================

# NETTOYAGE DU FICHIER ERP
df_erp_nettoye = nettoyer_dataframe(df_erp)

print("=== 1. ANALYSE DE LA VARIABLE PRIX (price) ===")
nb_missing_price = df_erp_nettoye["price"].isna().sum()
min_price = df_erp_nettoye["price"].min()
max_price = df_erp_nettoye["price"].max()

print(f"- Nombre d'articles avec un prix non renseigné : {nb_missing_price}")
print(f"- Prix minimum de la colonne price : {min_price:.2f} €")
print(f"- Prix maximum de la colonne price : {max_price:.2f} €")

# Isolation rigoureuse des prix négatifs (anomalies / erreurs de saisie)
prix_negatifs = df_erp_nettoye.loc[df_erp_nettoye["price"] < 0, :].copy()
print(f"- Nombre de prix négatifs détectés : {len(prix_negatifs)}")
display(prix_negatifs[['product_id', 'price', 'stock_quantity', 'stock_status']])


print("\n=== 2. ANALYSE DE LA VARIABLE STOCK (stock_quantity) ===")
min_stock = df_erp_nettoye["stock_quantity"].min()
max_stock = df_erp_nettoye["stock_quantity"].max()

print(f"- Quantité minimum en stock : {min_stock}")
print(f"- Quantité maximum en stock : {max_stock}")

# Isolation des stocks négatifs (commandes en souffrance / surventes)
stocks_negatifs = df_erp_nettoye.loc[df_erp_nettoye["stock_quantity"] < 0, :].copy()
print(f"- Nombre de stocks négatifs détectés : {len(stocks_negatifs)}")
display(stocks_negatifs[['product_id', 'price', 'stock_quantity', 'stock_status']])


print("\n=== 3. ANALYSE DE LA DISPONIBILITÉ WEB (onsale_web) ===")
print(f"Valeurs uniques de la colonne 'onsale_web' : {df_erp_nettoye['onsale_web'].unique()}")

# Analyse croisée rigoureuse : Stock dormant vs Rupture en ligne
stock_pas_en_ligne = df_erp_nettoye.loc[(df_erp_nettoye["stock_quantity"] > 0) & (df_erp_nettoye["onsale_web"] == 0)].copy()
stock_pas_en_ligne["erreur_type"] = "Stock dormant (en stock mais non publié)"

sans_stock_en_ligne = df_erp_nettoye.loc[(df_erp_nettoye["stock_quantity"] <= 0) & (df_erp_nettoye["onsale_web"] == 1)].copy()
sans_stock_en_ligne["erreur_type"] = "Survente / Rupture publiée (en ligne sans stock)"

df_anomalies_web = pd.concat([stock_pas_en_ligne, sans_stock_en_ligne])
print(f"-> Total des incohérences de publication web détectées : {len(df_anomalies_web)}")
print(f"   • Dont Stock dormant : {len(stock_pas_en_ligne)}")
print(f"   • Dont Rupture publiée : {len(sans_stock_en_ligne)}")


print("\n=== 4. ANALYSE DU PRIX D'ACHAT (purchase_price) ===")
print(f"- Nombre d'observations : {len(df_erp_nettoye)}")
print(f"- Nombre de valeurs manquantes : {df_erp_nettoye['purchase_price'].isna().sum()}")
print(f"- Prix d'achat min : {df_erp_nettoye['purchase_price'].min():.2f} €")
print(f"- Prix d'achat max : {df_erp_nettoye['purchase_price'].max():.2f} €")

# Contrôle programmatique de sécurité (Assertion)
assert df_erp_nettoye['purchase_price'].isna().sum() == 0, "Erreur : Valeurs manquantes détectées dans purchase_price !"
print("-> Vérification purchase_price validée : aucune valeur manquante.")
```

```python
# Eliminer les erreurs de prix et les stocks négatifs
df_erp_clean = df_erp_nettoye[(df_erp_nettoye['price'] >= 0) & (df_erp_nettoye['stock_quantity'] >= 0)].copy()

# Info erp après nettoyage
df_erp_clean.info()
```

```python
# Téléchargement du fichier WEB
df_web = charger_fichier("Sources/web.xlsx")

# Infos fichier Web
print(inspecter_dataframe(df_web))

# Nettoyage du fichier
df_web_nettoye = nettoyer_dataframe(df_web)
```

```python
# =========================================
# TRAITEMENT METIER SPECIFIQUE WEB NETTOYE
# =========================================

# Détection des SKU contenant des caractères spéciaux ou des formats non standards
mask_speciaux = df_web_nettoye['sku'].astype(str).str.contains(r'[-_]', regex=True)
intrus = df_web_nettoye[mask_speciaux].copy()

print(f"Nombre de SKU atypiques/spéciaux détectés : {len(intrus)}")
print("Détail des SKU spéciaux :")
display(intrus[['sku', 'post_title']])

# 1. Analyse de la colonne SKU (Clé de liaison avec l'ERP) sur le dataset nettoyé
nb_sku_vides = df_web_nettoye['sku'].isna().sum()
print(f"-> Nombre de lignes avec un SKU manquant (NaN) : {nb_sku_vides}")

# 2. Traitement des lignes sans SKU mais contenant des données de vente (Avoirs/Retours)
lignes_speciales_mask = df_web_nettoye['sku'].isna() & df_web_nettoye['post_title'].notna()
indices_speciaux = df_web_nettoye[lignes_speciales_mask].index.tolist()

for idx in indices_speciaux:
    df_web_nettoye.loc[idx, 'sku'] = f"avoir_inconnu_{idx}"
print(f"-> SKU fictifs affectés pour traçabilité sur les indices : {indices_speciaux}")

# 3. Suppression des lignes dont le SKU est manquant
df_web_nettoye = df_web_nettoye.dropna(subset=['sku']).copy()

# 4. Filtrage des fiches produits uniquement (élimination des pièces jointes / attachments)
print(f"\nRépartition de post_type avant filtrage :\n{df_web_nettoye['post_type'].value_counts(dropna=False)}")

df_web_clean = df_web_nettoye[df_web_nettoye['post_type'] == 'product'].copy()
df_web_clean = df_web_clean.reset_index(drop=True)

print(f"\nDimensions finales de df_web_clean : {df_web_clean.shape[0]} lignes, {df_web_clean.shape[1]} colonnes")
df_web_clean.info()
```

```python
# ==========================================================
# TRAITEMENT MÉTIER SPÉCIFIQUE : ANALYSE DE LA TABLE LIAISON
# ==========================================================

print("=" * 60)
print(" RAPPORT D'ANALYSE MÉTIER - TABLE DE LIAISON")
print("=" * 60)

# 1. Analyse de la colonne 'product_id' (Clé ERP)
nb_doublons_prod = df_liaison_nettoye['product_id'].duplicated().sum()
nb_nan_prod = df_liaison_nettoye['product_id'].isna().sum()
total_lignes = len(df_liaison_nettoye)

print(f"• Colonne 'product_id' (ERP) :")
print(f"  - Doublons détectés : {nb_doublons_prod} / {total_lignes} (Unicité garantie : {nb_doublons_prod == 0})")
print(f"  - Valeurs manquantes : {nb_nan_prod}")

print("-" * 60)

# 2. Analyse de la colonne 'id_web' (Clé Web / SKU)
nb_doublons_web = df_liaison_nettoye['id_web'].duplicated().sum()
nb_nan_web = df_liaison_nettoye['id_web'].isna().sum()

print(f"• Colonne 'id_web' (Web) :")
print(f"  - Doublons détectés : {nb_doublons_web} / {total_lignes}")
print(f"  - Valeurs manquantes (NaN) : {nb_nan_web}")

print("-" * 60)

# 3. Isolation et qualification métier des produits non vendus en ligne
lignes_id_web_vides = df_liaison_nettoye[df_liaison_nettoye['id_web'].isna()].copy()
nb_produits_erp_hors_web = lignes_id_web_vides['product_id'].nunique()

print(f"• Analyse des ruptures de liaison :")
print(f"  - Nombre de produits présents dans l'ERP mais absents du Web : {nb_produits_erp_hors_web}")
print("  -> Interprétation : Ce sont des références physiques gérées en stock mais non référencées sur le site e-commerce.")
print("=" * 60)
```

```python
# ========================================================================================================
# IDENTIFICATION DES OUTLIERS EN PRIX : Méthodes Z-SCORE (Z_indexe)  &  l'intervalle interquartile (IQR)
# ========================================================================================================

# --- 1. Méthode Z-Score ---
moyenne_prix = df_final['price'].mean()
ecart_t_prix = df_final['price'].std()

z_score_prix = (df_final['price'] - moyenne_prix) / ecart_t_prix
seuil_z3 = moyenne_prix + (3 * ecart_t_prix)

outliers_zscore = df_final[df_final['price'] > seuil_z3].copy()
outliers_zscore['z_score'] = round((outliers_zscore['price'] - moyenne_prix) / ecart_t_prix, 2)

print(f"IDENTIFICATION DES OUTLIERS EN PRIX : Méthode Z-SCORE :")
print(f"• Seuil de prix pour Z-score > 3 : {seuil_z3:.2f}€")
print(f"• Nombre d'outliers (Z-score) : {len(outliers_zscore)} produits")
print(f"• Proportion dans le catalogue : {(len(outliers_zscore) / len(df_final))*100: .2f}%\n" )


# --- 2. Méthode IQR (Intervalle Interquartile) ---
q1 = df_final['price'].quantile(0.25)
q3 = df_final['price'].quantile(0.75)
iqr = q3 - q1
seuil_max_iqr = q3 + 1.5 * iqr

# calcul des seuils
seuil_min = q1 - 1.5 * iqr
seuil_max = q3 + 1.5 * iqr

outliers_iqr = df_final[df_final['price'] > seuil_max_iqr].copy()

print()
print(f"IDENTIFICATION DES OUTLIERS EN PRIX : Méthode IQR :")
print(f"• Le seuil minimum du prix = {seuil_min: .2f}€")
print(f"• Le seuil maximum du prix = {seuil_max: .2f}€")
print(f"• Nombre d'outliers (IQR) : {len(outliers_iqr)} produits sur {len(df_final)}")
print(f"• Proportion dans le catalogue : {(len(outliers_iqr) / len(df_final)) * 100:.2f}%")

# Affichage du Top des produits de prestige (triés par prix décroissant)
print("\n Aperçu des 10 principaux produits outliers par prix :")
display(outliers_iqr[['product_id', 'post_title', 'price', 'purchase_price']].sort_values(by='price', ascending=False).head(10))


```

<div style="border: 1.5px solid black; padding: 10px;">
<h4  style="color:Blue;"> Comparaison détéction des outliers via les deux méthodes </h4>
<ul>
    <li><b>Méthode Z-Score :</b> Identifie 13 produits (soit 1,82% du catalogue) dépassant le seuil de 115,12 €.</li>
    <li><b>Méthode IQR : </b> Identifie 31 produits (soit 4,34% du catalogue) au-dessus du seuil de 84,09 € </li>
    <li><b>Diagnostic final :</b> on constate que la méthode Z-score est très sensible aux valeurs extrêmes (prix = 225€) ce qui tire la moyenne vers le haut tandis que la méthode IQR est plus reboste car elle s'appuie sur la médiane ce qui explique l'ecart du nombre d'outliers de 18 produits (31-13) entre les deux méthodes </li>
</ul>

</div>

```python
# ========================================================================================================
# ANALYSE DES OUTLIERS EN PRIX : Calcul du coefficient commercial et qualification de la marge
# ========================================================================================================

#Selon vous, ces outliers sont-ils justifiés ? Comment le démontrer si cela est possible ?
# 1. Calcul du coefficient commercial sur les outliers de prix
outliers_price_eval = outliers_iqr.copy()
outliers_price_eval["coef_commercial"] = (
    outliers_price_eval["price"] / outliers_price_eval["purchase_price"]
).round(2)

# 2. Fonction de qualification métier du coefficient
def commentaire_coef(coef):
    if coef < 1:
        return "⚠️ Vente à perte"
    elif 1 <= coef < 1.5:
        return "🔶 Marge faible"
    elif 1.5 <= coef < 3:
        return "✅ Marge normale"
    elif 3 <= coef < 5:
        return "⭐ Marge élevée"
    else:  # coef >= 5
        return "💎 Prix potentiellement abusif / Luxe"

# 3. Application de la fonction
outliers_price_eval["commentaire"] = outliers_price_eval["coef_commercial"].apply(commentaire_coef)

# 4. Affichage structuré et trié par coefficient décroissant
colonnes_affichage = ["product_id", "post_title", "price", "purchase_price", "coef_commercial", "commentaire"]
print(f"📊 Analyse de la rentabilité pour les {len(outliers_price_eval)} produits outliers (Méthode IQR) :\n")

# Affichage avec Pandas display pour une présentation élégante dans Jupyter
display(
    outliers_price_eval[colonnes_affichage]
    .sort_values(by="coef_commercial", ascending=False)
    .reset_index(drop=True)
)
```

```python
#=====================================
# Calculer le CA du site web 
#======================================

#Créer une colonne calculant le CA par article
df_final['CA'] = df_final['total_sales'] * df_final["price"]

#Calculer la somme de la colonne "ca_par_article"
total_CA = df_final['CA'].sum()

#Ce résultat correspond au chiffre d'affaire du site web
print(f" 💰 Chiffre d'affaires total du catalogue : {total_CA: .2f}€ \n")
```

```python
# ==========================================================
# ÉTAPE 5 - CALCUL PARETO (20/80) & COURBE DE LORENZ
# ==========================================================

# Tri et calculs
df_pareto = df_final.copy()
df_pareto = df_pareto.sort_values(by='CA', ascending=False).reset_index(drop=True)
df_pareto['Part CA'] = (df_pareto['CA'] / total_CA) * 100
df_pareto['cumul-part-ca'] = df_pareto['Part CA'].cumsum()

# 2. Analyse quantitative (Articles représentant 80% du CA)
articles_80 = df_pareto[df_pareto['cumul-part-ca'] <= 80]
nb_articles_80 = len(articles_80)
prop_articles_80 = (nb_articles_80 / len(df_pareto)) * 100

print(f"• Nombre d'articles représentant 80% du CA : {nb_articles_80} sur {len(df_pareto)} articles")
print(f"• Proportion de ces articles : {prop_articles_80:.2f}% \n")

# 3. Graphique Pareto / Lorenz
n_articles = len(df_pareto)
x_values = np.arange(1, n_articles + 1) / n_articles * 100
y_values = df_pareto['cumul-part-ca'].values

fig, ax = plt.subplots(figsize=(10, 6))

# Courbes principales
ax.plot(x_values, y_values, color='#006400', linewidth=3, label='Courbe de répartition réelle')
ax.plot([0, 100], [0, 100], color='grey', linestyle='--', label='Ligne d’égalité parfaite')

# Point d'inflexion (ex: ~56% des articles génèrent 80% du CA)
point_x = prop_articles_80
ax.axvline(x=point_x, color='red', linestyle=':', alpha=0.6)
ax.axhline(y=80, color='red', linestyle=':', alpha=0.6)
ax.scatter(point_x, 80, color='red', s=100, zorder=5, label=f'Notre point : {point_x:.2f}% articles = 80% CA')

# Habillage
ax.set_title('Courbe de Lorenz : Analyse de la concentration du CA', fontsize=14, fontweight='bold', pad=20)
ax.set_xlabel('% cumulé des articles', fontsize=12)
ax.set_ylabel('% cumulé du CA', fontsize=12)
ax.set_xlim(0, 100)
ax.set_ylim(0, 100)
ax.legend(loc='upper left')
ax.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()

# Export & Affichage
plt.savefig("Graphiques/CA_Pareto.png", dpi=300, bbox_inches="tight")
plt.show()
```

</div>

```python
# =======================================================
# ANALYSE PARETO EN QUANTITÉ (Volumes)
# =======================================================

# 1. Calcul de la part de chaque ligne en quantité
df_quantites['Part-quantité'] = df_quantites['total_sales'] / df_quantites['total_sales'].sum()

# 2. Tri décroissant et calcul du cumul (de 0 à 1.0)
df_quantites = df_quantites.sort_values(by='total_sales', ascending=False).reset_index(drop=True)
df_quantites['Cumul-part_quantité'] = df_quantites['Part-quantité'].cumsum()

# 3. Filtrage : on cherche les lignes qui constituent les premiers 80% (0.8)
qte_80 = df_quantites[df_quantites['Cumul-part_quantité'] <= 0.8]

# 4. RÉSULTATS
nbr_produit_80 = len(qte_80)                    # Nombre de produits différents
volume_ventes_80 = qte_80['total_sales'].sum()    # Somme des quantités de ces produits
prop_catalogue = (nbr_produit_80 / len(df_quantites)) * 100

print(f"- Nombre de produits nécessaires : {nbr_produit_80} sur {len(df_quantites)} produits")
print(f"- Volume des ventes (80% cible) : {volume_ventes_80:,.0f} unités")
print(f"- Proportion de ces articles par rapport au catalogue : {prop_catalogue:.2f}%")
```

<div style="border: 1px solid RGB(51,165,182);" >
<h3 style="margin: auto; padding: 20px; color: RGB(51,165,182); ">Etape 5.3 - Analyse des stocks</h3>
</div>

```python
# ========================================================================================================
# ANALYSE DES STOCKS : ROTATION & VALORISATION FINANCIÈRE (VERSION V2 CORRIGÉE)
# ========================================================================================================

# 1. Calcul de la rotation de stock (en mois)
# On gère les divisions par zéro ou valeurs infinies avec .replace() et .fillna()
df_final['Rotation de stock'] = round(df_final['stock_quantity'] / df_final['total_sales'], 2)
df_final['Rotation de stock'] = df_final['Rotation de stock'].replace([np.inf, -np.inf], np.nan).fillna(0)

# 2. Tri décroissant pour identifier les "Flops" (produits encombrants / stocks dormants)
df_rotation_stock = df_final.sort_values(by='Rotation de stock', ascending=False).reset_index(drop=True)

# 3. Préparation du Flop 20
flop_20 = df_rotation_stock.head(20)

# 4. Création du graphique Matplotlib pour le Flop 20
fig, ax = plt.subplots(figsize=(12, 8))

# Troncature propre des noms de produits
raw_names = flop_20['post_title'].astype(str).values
clean_names = [n.replace('-', ' ')[:38] + '...' if len(n) > 38 else n.replace('-', ' ') for n in raw_names]

# Création des barres horizontales
bars = ax.barh(np.arange(len(clean_names)), flop_20['Rotation de stock'], color='#e74c3c', edgecolor='white', linewidth=0.5)

# Application explicite des noms sur l'axe Y
ax.set_yticks(np.arange(len(clean_names)))
ax.set_yticklabels(clean_names, fontsize=9)

# Ajout des valeurs numériques au bout des barres (avec gestion pour éviter les superpositions)
for bar in bars:
    width = bar.get_width()
    ax.text(width + 0.5, 
            bar.get_y() + bar.get_height()/2, 
            f'{width:,.1f} mois', 
            va='center', ha='left', 
            fontsize=8.5, fontweight='bold', color='#333333')

# Esthétique des axes et du titre
ax.set_title('Flop 20 : Produits avec la plus forte rotation de stock (Stock dormant)', fontsize=14, pad=20, fontweight='bold')

# Nettoyage des bordures
for spine in ['top', 'right']:
    ax.spines[spine].set_visible(False)

# Marge à droite pour les étiquettes et inversion pour avoir le pire en haut
ax.set_xlim(0, flop_20['Rotation de stock'].max() + 5)
ax.invert_yaxis()

# ⚠️ AJUSTEMENT CLÉ : Marge de gauche élargie (35%) pour que les noms longs ne soient jamais coupés
plt.subplots_adjust(left=0.35, right=0.90, top=0.90, bottom=0.1)

# Export du graphique
plt.savefig("Graphiques/rotation_stock.png", dpi=300, bbox_inches="tight")
plt.show()

# ========================================================================================================
# 5. VALORISATION FINANCIÈRE ET VOLUMÉTRIQUE DES STOCKS
# ========================================================================================================

# Valorisation en euros (Quantité restante × Prix d'achat)
df_final['Valorisation_stock_euros'] = df_final['stock_quantity'] * df_final['purchase_price']
valeur_totale_stock = df_final['Valorisation_stock_euros'].sum()

# Volume total de bouteilles en stock
volume_total_stock = df_final['stock_quantity'].sum()

# Affichage propre des indicateurs clés de stock
print(f"📦 Volume total en stock : {volume_total_stock:,.0f} bouteilles")
print(f"💰 Valeur totale du stock immobilisé : {valeur_totale_stock:,.2f} €")
```

<div style="border: 1.5px solid black; padding: 10px;">
    <h4  style="color:Blue;">Analyse valorisation des stocks en euros</h4>
    <p> La valeur de stock à hauteur de 298,540.07€ est un chiffre impressionnant. C'est de l'argent qui dorment et ne peut pas être investi ailleurs (nouveaux produits, marketing, etc.) tant que les  17803 bouteilles de vin ne sont pas vendues.</p>
    <p>La catégorie Champagnes domine le Flop20 et montre que le déstockage de ces produits se fait entre 14 et 31 mois ce qui laisse le capital investi ne circule pas assez vite.</p>

</div>


<div style="border: 1px solid RGB(51,165,182);" >
<h3 style="margin: auto; padding: 20px; color: RGB(51,165,182); ">Etape 5.4 - Analyse du taux de marge</h3>
</div>

```python
# ========================================================================================================
# ÉTAPE 7 - ANALYSE DE LA RENTABILITÉ : TAUX DE MARGE (VERSION V2)
# ========================================================================================================

print("=" * 60)
print(" ANALYSE DES TAUX DE MARGE ET DE LA RENTABILITÉ")
print("=" * 60)

# 1. Création des colonnes de prix HT
# On suppose une TVA à 20% (diviseur 1.2) pour le prix de vente et le prix d'achat
df_final['prix_ht'] = df_final['price'] / 1.2
df_final["Prix_achat_HT"] = df_final["purchase_price"] / 1.2

# 2. Calcul du taux de marge global
# Formule : ((Prix de vente HT - Prix d'achat HT) / Prix d'achat HT) * 100
df_final['taux_marge'] = ((df_final['prix_ht'] - df_final['Prix_achat_HT']) / df_final['Prix_achat_HT']) * 100

# 3. Affichage des bornes globales
print(f"• Le taux de marge minimum global est de : {df_final['taux_marge'].min():.4f}%")
print(f"• Le taux de marge maximum global est de : {df_final['taux_marge'].max():.4f}%")

# 4. Identification des anomalies (produits vendus à perte / taux de marge < 0)
print("\n- Vérification des produits avec un taux de marge négatif (< 0) :")
anomalies_marge = df_final.loc[(df_final['taux_marge'] < 0), ['product_id', 'price', 'stock_quantity', 'stock_status', 'purchase_price', 'taux_marge']]
print(anomalies_marge if not anomalies_marge.empty else "Aucun produit avec une marge négative.")

# 5. Création d'un dataframe filtré avec les taux de marge strictement positifs
df_marge_positive = df_final.loc[(df_final['taux_marge'] > 0), :].copy()

min_marge_positive = df_marge_positive['taux_marge'].min()
max_marge_positive = df_marge_positive['taux_marge'].max()

print(f"\n• Le taux de marge minimum (positif) est de : {min_marge_positive:.4f}%")
print(f"• Le taux de marge maximum (positif) est de : {max_marge_positive:.4f}%")

# 6. Agrégation par type de produit (Moyenne du taux de marge)
df_marge_par_type = df_marge_positive.groupby('product_type')['taux_marge'].mean().reset_index()
df_marge_par_type = df_marge_par_type.sort_values(by='taux_marge', ascending=False)

print("\n- Taux de marge moyen par catégorie de produit : ")
print(df_marge_par_type)
print("=" * 60)

# 7. Visualisation graphique du taux de marge moyen par type de produit
plt.figure(figsize=(10, 6))

sns.barplot(
    x='taux_marge', 
    y='product_type', 
    data=df_marge_par_type, 
    hue='product_type', 
    palette='viridis', 
    legend=False
)

plt.title('- Taux de marge moyen par catégorie de produit (%)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Taux de marge moyen', fontsize=11, fontweight='bold')
plt.ylabel('Catégorie', fontsize=11, fontweight='bold')
plt.grid(axis='x', linestyle='--', alpha=0.6)

# Nettoyage des bordures
sns.despine(top=True, right=True)

# Ajout des étiquettes de score sur les barres
for index, value in enumerate(df_marge_par_type['taux_marge']):
    plt.text(value + 0.5, index, f' {value:.2f}%', va='center', fontsize=9, fontweight='bold')

# Ajustement de l'axe X pour laisser de la place aux étiquettes
plt.xlim(0, df_marge_par_type['taux_marge'].max() + 10)

# Export graphique
plt.savefig("Graphiques/marge_cat_produit.png", dpi=300, bbox_inches="tight")
plt.show()
```

<div style="border: 1.5px solid black; padding: 10px;">    
<h4  style="color:Blue;">Taux de marge moyen par catégorie de produit (%) </h4>
<ul>
    <li>Le Cognac et le Whisky affichent les taux de marge moyenne les plus élevés (autour de 118-119%). </li>
    <li>Le Gin et le Vin suivent de près (109% et 93%).</li>
    <li>Le Champagne et l'Huile d'olive se positionnent plus bas (autour de 60-67%), ce qui est cohérent avec des dynamiques de rotation de stock ou des positionnements de prix différents.</li>
</ul>

</div>

```python
#Que peut-on conclure des corrélations ?
# ========================================================================================================
# ANALYSE DES CORRÉLATIONS (HEATMAP TRIANGULAIRE) avec le dataframe commercial
# ========================================================================================================

print("=" * 70)
print(" GÉNÉRATION DE LA MATRICE DE CORRÉLATION")
print("=" * 70)

# 1. Sélection sécurisée des variables clés
cols_corr = ['stock_quantity', 'total_sales', 'price', 'taux_marge', 'purchase_price', 'Rotation de stock']

# Filtrage automatique pour éviter un plantage si une colonne est mal orthographiée
cols_disponibles = [col for col in cols_corr if col in df_marge_positive.columns]
df_corr = df_marge_positive[cols_disponibles].corr()

# 2. Création du masque pour n'afficher que la moitié inférieure de la matrice
mask = np.triu(np.ones_like(df_corr, dtype=bool))

# 3. Configuration de la figure
plt.figure(figsize=(10, 8))

# 4. Création de la heatmap
sns.heatmap(
    df_corr, 
    mask=mask, 
    annot=True, 
    fmt=".2f", 
    cmap='coolwarm', 
    vmin=-1, vmax=1, 
    center=0,
    square=True, 
    linewidths=.5,
    cbar_kws={"shrink": .8}
)

plt.title('Matrice de corrélation : Stocks, Ventes et Prix', fontsize=14, fontweight='bold', pad=15)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)

# 5. Sécurisation et export du graphique
os.makedirs("Graphiques", exist_ok=True)
output_path = "Graphiques/corr_stock_ventes_prix.png"
plt.savefig(output_path, dpi=300, bbox_inches="tight")

print(f"✅ Matrice de corrélation générée et sauvegardée avec succès sous : {output_path}")

plt.show()
```

<div style="border: 1.5px solid black; padding: 10px;">
    <h4  style="color:Blue;">Les corrélations :</h4>    
    <p> <span style= "font-weight:bold;">- Corrélation positive :</span> entre stock_quantity et rotation de stock. Cela confirme que le "Flop 20" est principalement causé par un volume d'unités trop élevé par rapport à la vitesse de vente</p>   
    <p><span style= "font-weight:bold;">- Corrélation négative :</span>  Elle indique que plus le prix (price) est élevé, plus les ventes totales (total_sales) ont tendance à baisser. C'est un argument de poids pour justifier une baisse de prix sur les Champagnes qui dorment en stock.</p>

<p><span style= "font-weight:bold;">- Une corrélation presque parfaite : </span>  entre purchase_price et price. Cela montre que la politique de prix est très rigide : les prix de vente sont fixés directement en fonction des coûts d'achat, sans forcément prendre en compte la demande du marché (ce qui explique parfois les stocks qui stagnent)</p>
