```python
#==========================================
# 1) TELECHARGEMENT DES FICHIERS
#==========================================

def charger_fichier(chemin_fichier: str) -> pd.DataFrame:
    """
    Charge un fichier Excel de manière sécurisée.
    """
    try:
        df = pd.read_excel(chemin_fichier)
        print(f"📄 Fichier '{chemin_fichier}' chargé : {df.shape[0]} lignes, {df.shape[1]} colonnes.")
        return df
    except Exception as e:
        print(f"❌ Erreur lors du chargement de {chemin_fichier} : {e}")
        return None


# ==========================================
# 2) LECTURE ET AFFICHAGE DÉTAILLÉ DES FICHIERS
# ==========================================

def lecture_dataframe(df: pd.DataFrame, nom_df: str):
    """
    Affiche le rapport détaillé pour un fichier donné :
    1. Résumé structurel (.info())
    2. Aperçu des 3 premières lignes (sans troncature)
    """
    print("=" * 60)
    print(f" RAPPORT DÉTAILLÉ : {nom_df}")
    print("=" * 60)
    
    # 1. Info du DataFrame
    print("\n[1] Résumé structurel (.info()) :")
    df.info()
    
    # 2. Affichage des 3 premières lignes
    print("\n[2] Aperçu des 3 premières lignes :")
    with pd.option_context('display.max_rows', None, 'display.max_columns', None):
        display(df.head(3))
  
    
#==========================================
# 3) INSPECTION DES FICHIERS
#==========================================
    
def inspecter_dataframe(df: pd.DataFrame, nom_df: str) -> dict:
    """
    Analyse un DataFrame et retourne un dictionnaire de synthèse 
    regroupant l'intégrité globale et les statistiques de valeurs manquantes.
    """
    nb_lignes, nb_cols = df.shape
    nb_doublons = df.duplicated().sum()
    
    # Vérifications d'intégrité
    lignes_vides = df.isna().all(axis=1).sum()
    cols_vides = df.isna().all(axis=0).sum()
    
    df_num = df.select_dtypes(include=[np.number])
    lignes_zeros = (df_num == 0).all(axis=1).sum() if not df_num.empty else 0
    cols_zeros = (df_num == 0).all(axis=0).sum() if not df_num.empty else 0
    
    # Valeurs manquantes totales
    total_na = df.isna().sum().sum()
    
    return {
        'Source': nom_df,
        'Lignes': nb_lignes,
        'Colonnes': nb_cols,
        'Doublons': nb_doublons,
        'Lignes Vides': lignes_vides,
        'Cols Vides': cols_vides,
        'Lignes Zéros': lignes_zeros,
        'Cols Zéros': cols_zeros,
        'Total Valeurs Manquantes': total_na,
        'Statut': 'Audité'
    }


#==========================================
# 4) NETTOYAGE DES FICHIERS
#==========================================

def nettoyer_dataframe(df: pd.DataFrame, 
                       cols_protegees: list = None, 
                       supprimer_constantes: bool = False,
                       supprimer_zeros: bool = False) -> pd.DataFrame:
    """
    Applique un nettoyage standardisé sécurisé sur un DataFrame :
    1. Suppression des doublons
    2. Suppression des lignes entièrement vides
    3. Suppression des colonnes entièrement vides (NaN)
    4. Suppression optionnelle des colonnes de zéros
    5. Suppression optionnelle des colonnes constantes (avec protection possible)
    6. Séparation des colonnes datetime en Date et Heure
    """
    if cols_protegees is None:
        cols_protegees = []
        
    print("=" * 60)
    print(" EXÉCUTION DU PIPELINE DE NETTOYAGE SÉCURISÉ")
    print("=" * 60)
    print(f"Dimensions avant nettoyage : {df.shape[0]} lignes | {df.shape[1]} colonnes")
    
    df_clean = df.copy()
    
    # 1. Suppression des doublons
    nb_doublons = df_clean.duplicated().sum()
    df_clean = df_clean.drop_duplicates()
    print(f"✓ Doublons supprimés : {nb_doublons}")
    
    # 2. Suppression des lignes entièrement vides
    lignes_vides = df_clean.isna().all(axis=1).sum()
    df_clean = df_clean.dropna(how='all')
    print(f"✓ Lignes entièrement vides supprimées : {lignes_vides}")
    
    # 3. Suppression des colonnes entièrement vides (NaN)
    cols_vides = df_clean.columns[df_clean.isna().all()].tolist()
    df_clean = df_clean.dropna(axis=1, how='all')
    print(f"✓ Colonnes vides (NaN) supprimées : {cols_vides}")     
             
    print(f"Dimensions finales après nettoyage : {df_clean.shape[0]} lignes | {df_clean.shape[1]} colonnes")
    print("-" * 60 + "\n")
    
    return df_clean


#============================================================
# 5) COMPARAISON DE DONNEES AVANT ET APRES LE NETTOYANGE
#=============================================================

def generer_tableau_comparatif(dict_avant: dict, dict_apres: dict):
    """
    Génère les métriques comparatives avant et après nettoyage et 
    affiche le résultat sous forme de graphique stylisé avec une première colonne élargie.
    """
    comparaison_data = []
    
    for nom in dict_avant.keys():
        df_av = dict_avant[nom]
        df_ap = dict_apres[nom]
        
        lignes_av, cols_av = df_av.shape
        doublons_av = df_av.duplicated().sum()
        vides_av = df_av.isna().sum().sum()
        
        lignes_ap, cols_ap = df_ap.shape
        doublons_ap = df_ap.duplicated().sum()
        vides_ap = df_ap.isna().sum().sum()
        
        comparaison_data.append({
            'Source': nom,
            'Lignes\n(Avant)': lignes_av,
            'Lignes\n(Après)': lignes_ap,
            'Δ Lignes': lignes_ap - lignes_av,
            'Colonnes\n(Avant)': cols_av,
            'Colonnes\n(Après)': cols_ap,
            'Doublons\n(Avant)': doublons_av,
            'Doublons\n(Après)': doublons_ap,
            'Manquantes\n(Avant)': vides_av,
            'Manquantes\n(Après)': vides_ap,
            'Statut': 'Nettoyé'
        })
        
    df_comparaison = pd.DataFrame(comparaison_data)
    
    fig, ax = plt.subplots(figsize=(15, 3.5))
    ax.axis('off')
    ax.axis('tight')

    table = ax.table(
        cellText=df_comparaison.values,
        colLabels=df_comparaison.columns,
        cellLoc='center',
        loc='center'
    )

    table.auto_set_font_size(False)
    table.set_fontsize(9.5)
    table.scale(1.2, 2.2)

    # Ajustement personnalisé de la largeur des colonnes
    for (row, col), cell in table.get_celld().items():
        cell.set_edgecolor('#dcdcdc')
        
        # Élargir spécifiquement la première colonne (colonne 0)
        if col == 0:
            cell.set_width(0.18)
            
        if row == 0:
            cell.set_facecolor('#2c3e50')
            cell.set_text_props(color='white', weight='bold')
        else:
            if row % 2 == 0:
                cell.set_facecolor('#f8f9fa')
            else:
                cell.set_facecolor('#ffffff')
                
            if col == 3:
                valeur_delta = df_comparaison.iloc[row - 1]['Δ Lignes']
                if valeur_delta < 0:
                    cell.set_text_props(color='#e74c3c', weight='bold')

    plt.title("TABLEAU DE BORD EXÉCUTIF — IMPACT DU NETTOYAGE", fontsize=12, weight='bold', pad=8)
    plt.savefig('Graphiques/tableau_comparatif_stylise.png', bbox_inches='tight', dpi=300)
    plt.show()
    
    return df_comparaison
    
```

```python
# Correction et vérification du statut du stock dans l'ERP
# Règle : stock > 0 -> 'instock', stock <= 0 -> 'outofstock'

# 1. Identification des incohérences avant correction
mask_incoherents = (
    ((df_erp['stock_quantity'] > 0) & (df_erp['stock_status'] == 'outofstock')) |
    ((df_erp['stock_quantity'] <= 0) & (df_erp['stock_status'] == 'instock'))
)
print(f"Nombre d'erreurs de statut détectées : {mask_incoherents.sum()}")

# Affichage des lignes concernées
if mask_incoherents.sum() > 0:
    print("\n=== LIGNES AVEC STATUT DE STOCK INCOHÉRENT ===")
    display(df_erp[mask_incoherents])

# 2. Application de la correction vectorisée
df_erp['stock_status'] = np.where(
    df_erp['stock_quantity'] > 0, 
    'instock', 
    'outofstock'
)
print("✓ Statuts stock mis à jour avec succès.")
```

```python
# Correction et vérification du statut du stock dans l'ERP
# Règle : stock > 0 -> 'instock', stock <= 0 -> 'outofstock'

# Exemple : Exécuter une requête SQL pour trouver les erreurs de statut stock dans df_erp
query = """
    SELECT product_id, price, stock_quantity, stock_status
    FROM df_erp
    WHERE (stock_quantity > 0 AND stock_status = 'outofstock')
       OR (stock_quantity <= 0 AND stock_status = 'instock')
"""

df_erreurs_sql = duckdb.query(query).to_df()
display(df_erreurs_sql)

# 2. Application de la correction vectorisée
df_erp['stock_status'] = np.where(
    df_erp['stock_quantity'] > 0, 
    'instock', 
    'outofstock'
)
print("✓ Statuts stock mis à jour avec succès.")
```

```python
# ===================================
# 1) VERIFICATION DU STATUT DE STOCK
# ==================================

print(df_erp.info())

# Valeurs uniques de la colonne stock_status
print(f"- Les valeurs uniques de la variable stack_status : \n {df_erp['stock_status'].unique()}")

# 1. Détermination du statut théorique basé sur la quantité en stock . Règle métier : Si stock_quantity > 0 ==> instock, sinon outofstock
statut_theorique = np.where(df_erp['stock_quantity'] > 0, 'instock', 'outofstock')

# 2. Détection des incohérences de départ
masque_incoherence = df_erp['stock_status'] != statut_theorique
nb_incoherences_depart = masque_incoherence.sum()
nb_conformes_depart = len(df_erp) - nb_incoherences_depart

print(f"=== SITUATION DE DÉPART METIER ===")
print(f"- Nombre de lignes conformes stock_status au départ (True)  : {nb_conformes_depart} / {len(df_erp)}")
print(f"- Nombre de lignes INCOHÉRENTES stock_status au départ (False) : {nb_incoherences_depart} / {len(df_erp)}")

# 3. Affichage des lignes avec un statut incohérent au départ
print(f"\nDétail des {nb_incoherences_depart} anomalies détectées au départ :")
df_anomalies = df_erp[masque_incoherence].copy()
display(df_anomalies[['product_id', 'stock_quantity', 'stock_status']])

# 4. Correction directe du statut erroné
df_erp['stock_status'] = statut_theorique


# 5. Vérification post-correction (Contrôle programmatique)
anomalies_restantes = (df_erp['stock_status'] != np.where(df_erp['stock_quantity'] > 0, 'instock', 'outofstock')).sum()
assert anomalies_restantes == 0, "Erreur : Des incohérences subsistent après correction !"

print(f"\n=== BILAN POST-CORRECTION ===")
print(f"Nombre d'anomalies restantes : {anomalies_restantes}")
print("Le statut de stock du fichier ERP est désormais 100% cohérent avec les quantités réelles.")
```

```python

# 1. Calculs intermédiaires pour les incohérences web
stock_pas_en_ligne = df_erp.loc[(df_erp["stock_quantity"] > 0) & (df_erp["onsale_web"] == 0)].copy()
stock_pas_en_ligne["erreur_type"] = "Stock dormant (en stock mais non publié)"

sans_stock_en_ligne = df_erp.loc[(df_erp["stock_quantity"] <= 0) & (df_erp["onsale_web"] == 1)].copy()
sans_stock_en_ligne["erreur_type"] = "Survente / Rupture publiée (en ligne sans stock)"

df_anomalies_web = pd.concat([stock_pas_en_ligne, sans_stock_en_ligne])

# 2. Tableau récapitulatif global enrichi
summary_data = [
    {
        "Variable": "Prix (price)", 
        "Valeurs manquantes": df_erp["price"].isna().sum(), 
        "Min": f"{df_erp['price'].min():.2f} €", 
        "Max": f"{df_erp['price'].max():.2f} €", 
        "Anomalies / Erreurs": f"{(df_erp['price'] < 0).sum()} prix négatifs"
    },
    {
        "Variable": "Stock (stock_quantity)", 
        "Valeurs manquantes": df_erp["stock_quantity"].isna().sum(), 
        "Min": df_erp["stock_quantity"].min(), 
        "Max": df_erp["stock_quantity"].max(), 
        "Anomalies / Erreurs": f"{(df_erp['stock_quantity'] < 0).sum()} stocks négatifs"
    },
    {
        "Variable": "Disponibilité web (onsale_web)", 
        "Valeurs manquantes": df_erp["onsale_web"].isna().sum(), 
        "Min": "-", 
        "Max": "-", 
        "Anomalies / Erreurs": f"Total : {len(df_anomalies_web)} (Dormant : {len(stock_pas_en_ligne)} | Rupture : {len(sans_stock_en_ligne)})"
    },
    {
        "Variable": "Prix d'achat (purchase_price)", 
        "Valeurs manquantes": df_erp['purchase_price'].isna().sum(), 
        "Min": f"{df_erp['purchase_price'].min():.2f} €", 
        "Max": f"{df_erp['purchase_price'].max():.2f} €", 
        "Anomalies / Erreurs": 0
    }
]

df_summary = pd.DataFrame(summary_data)
print("=== TABLEAU DE SYNTHÈSE DES ANALYSES ===")
display(df_summary)

# 3. Isolation et affichage des listes d'anomalies
prix_negatifs = df_erp.loc[df_erp["price"] < 0, ['product_id', 'price', 'stock_quantity', 'stock_status']].copy()
stocks_negatifs = df_erp.loc[df_erp["stock_quantity"] < 0, ['product_id', 'price', 'stock_quantity', 'stock_status']].copy()

print("\n=== LISTE DES PRIX NÉGATIFS ===")
display(prix_negatifs)

print("\n=== LISTE DES QUANTITÉS EN STOCK NÉGATIVES ===")
display(stocks_negatifs)

print(f"\n=== DÉTAIL DES INCOHÉRENCES WEB ({len(df_anomalies_web)} anomalies) ===")
display(df_anomalies_web[['product_id', 'price', 'stock_quantity', 'stock_status', 'onsale_web', 'erreur_type']])

# Assertion de sécurité
assert df_erp['purchase_price'].isna().sum() == 0, "Erreur : Valeurs manquantes détectées dans purchase_price !"
```

```python
# Fusion en partant de l'ERP pour conserver l'intégralité du stock
df_intermediaire = pd.merge(
    df_erp_clean, 
    df_liaison, 
    on='product_id', 
    how='left'
)

df_final = pd.merge(
    df_intermediaire, 
    df_web_clean, 
    left_on='id_web', 
    right_on='sku', 
    how='left'
)
print(df_final.shape)
print(df_final.info())
```

<div style="border: 1.5px solid black; padding: 10px;">
    <h2  style="color:Blue;"> Analyse du merge final</h2>
<h3>La logique métier (ERP = Référentiel central) :</h3>
<ul>
    <li> l'ERP est la source de vérité pour les produits : chaque ligne de l'ERP représente un produit réel de votre stock physique.</li>
    <li> Avec <b>how='left'</b>  en partant de l'ERP (ou du merge ERP + Liaison), on conserve exactement les 820 produits de l'ERP </li>
    <li> Si un produit est présent dans l'ERP mais n'a pas encore de correspondance sur le site web, il garde ses informations ERP et aura des NaN pour les colonnes web</li>    
<li>Avec <b> how = 'outer' </b>, on récupère absolument toutes les lignes de tous les fichiers, y compris des références présentes dans le fichier web ou de liaison mais qui n'existent plus dans l'ERP, ce qui fausse le nombre total de produits (ex 825(v1) lignes au lieu de 820). </li>      
</ul>
</div>

```python
import duckdb

# DuckDB accède directement aux variables du namespace global (vos DataFrames Pandas)
# Exemple : Refaire la jointure ERP / Liaison / Web et calculer le chiffre d'affaires par produit

query = """
    SELECT 
        e.product_id,
        w.post_title,
        e.price,
        w.total_sales,
        (e.price * w.total_sales) AS CA_produit
    FROM df_erp_clean e
    LEFT JOIN df_liaison l ON e.product_id = l.product_id
    LEFT JOIN df_web_clean w ON l.id_web = w.sku
    WHERE w.total_sales > 0
    ORDER BY CA_produit DESC
    LIMIT 20;
"""

result_duckdb = duckdb.query(query).to_df()
display(result_duckdb)
```

</div>

```python
# ========================================================================================================
# IDENTIFICATION DES OUTLIERS EN PRIX : Méthodes Z-SCORE (Z_indexe)  &  l'intervalle interquartile (IQR)
# ========================================================================================================

# --- 1. Méthode Z-Score ---
moyenne_prix = df_final['price'].mean()
ecart_t_prix = df_final['price'].std()

z_score_prix = (df_final['price'] - moyenne_prix) / ecart_t_prix
seuil_z3 = moyenne_prix + (3 * ecart_t_prix)

outliers_zscore = df_final[df_final['price'] > seuil_z3].copy()
outliers_zscore['z_score'] = round((outliers_zscore['price'] - moyenne_prix) / ecart_t_prix, 2)

print(f"IDENTIFICATION DES OUTLIERS EN PRIX : Méthode Z-SCORE :")
print(f"• Seuil de prix pour Z-score > 3 : {seuil_z3:.2f}€")
print(f"• Nombre d'outliers (Z-score) : {len(outliers_zscore)} produits")
print(f"• Proportion dans le catalogue : {(len(outliers_zscore) / len(df_final))*100: .2f}%\n" )


# --- 2. Méthode IQR (Intervalle Interquartile) ---
q1 = df_final['price'].quantile(0.25)
q3 = df_final['price'].quantile(0.75)
iqr = q3 - q1
seuil_max_iqr = q3 + 1.5 * iqr

# calcul des seuils
seuil_min = q1 - 1.5 * iqr
seuil_max = q3 + 1.5 * iqr

outliers_iqr = df_final[df_final['price'] > seuil_max_iqr].copy()

print()
print(f"IDENTIFICATION DES OUTLIERS EN PRIX : Méthode IQR :")
print(f"• Le seuil minimum du prix = {seuil_min: .2f}€")
print(f"• Le seuil maximum du prix = {seuil_max: .2f}€")
print(f"• Nombre d'outliers (IQR) : {len(outliers_iqr)} produits sur {len(df_final)}")
print(f"• Proportion dans le catalogue : {(len(outliers_iqr) / len(df_final)) * 100:.2f}%")

# Affichage du Top des produits de prestige (triés par prix décroissant)
print("\n Aperçu des 10 principaux produits outliers par prix :")
display(outliers_iqr[['product_id', 'post_title', 'price', 'purchase_price']].sort_values(by='price', ascending=False).head(10))
```

```python
# ========================================================================================================
# ANALYSE DES OUTLIERS EN PRIX : Calcul du coefficient commercial et qualification de la marge
# ========================================================================================================

#Selon vous, ces outliers sont-ils justifiés ? Comment le démontrer si cela est possible ?
# 1. Calcul du coefficient commercial sur les outliers de prix
outliers_price_eval = outliers_iqr.copy()
outliers_price_eval["coef_commercial"] = (
    outliers_price_eval["price"] / outliers_price_eval["purchase_price"]
).round(2)

# 2. Fonction de qualification métier du coefficient
def commentaire_coef(coef):
    if coef < 1:
        return "⚠️ Vente à perte"
    elif 1 <= coef < 1.5:
        return "🔶 Marge faible"
    elif 1.5 <= coef < 3:
        return "✅ Marge normale"
    elif 3 <= coef < 5:
        return "⭐ Marge élevée"
    else:  # coef >= 5
        return "💎 Prix potentiellement abusif / Luxe"

# 3. Application de la fonction
outliers_price_eval["commentaire"] = outliers_price_eval["coef_commercial"].apply(commentaire_coef)

# 4. Affichage structuré et trié par coefficient décroissant
colonnes_affichage = ["product_id", "post_title", "price", "purchase_price", "coef_commercial", "commentaire"]
print(f"📊 Analyse de la rentabilité pour les {len(outliers_price_eval)} produits outliers (Méthode IQR) :\n")

# Affichage avec Pandas display pour une présentation élégante dans Jupyter
display(
    outliers_price_eval[colonnes_affichage]
    .sort_values(by="coef_commercial", ascending=False)
    .reset_index(drop=True)
)
```

```python
#=====================================
# Calculer le CA du site web 
#======================================

#Créer une colonne calculant le CA par article
df_final['CA'] = df_final['total_sales'] * df_final["price"]

#Calculer la somme de la colonne "ca_par_article"
total_CA = df_final['CA'].sum()

#Ce résultat correspond au chiffre d'affaire du site web
print(f" 💰 Chiffre d'affaires total du catalogue : {total_CA: .2f}€ \n")
```

```python
# ========================================================================================================
# ÉTAPE 5 - COURBE DE LORENZ & PARETO (Comparatif Global vs Produits Actifs)
# ========================================================================================================

# 1. Définition des deux sous-ensembles
datasets = {
    "Catalogue Global": df_final.copy(),
    "Produits Actifs (total_sales > 0)": df_final[df_final['total_sales'] > 0].copy()
}

# Création d'une figure avec 2 sous-graphiques côte à côte
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

for ax, (nom, df_subset) in zip(axes, datasets.items()):
    # Copie, tri et calculs Pareto
    df_pareto = df_subset.copy()
    
    # Sécurité si la colonne 'CA' n'est pas encore créée (CA = price * total_sales)
    if 'CA' not in df_pareto.columns:
        df_pareto['CA'] = df_pareto['price'] * df_pareto['total_sales']
        
    df_pareto = df_pareto.sort_values(by='CA', ascending=False).reset_index(drop=True)
    total_ca_subset = df_pareto['CA'].sum()
    
    # Évite la division par zéro si le CA total est nul
    if total_ca_subset > 0:
        df_pareto['Part CA'] = (df_pareto['CA'] / total_ca_subset) * 100
    else:
        df_pareto['Part CA'] = 0
        
    df_pareto['cumul-part-ca'] = df_pareto['Part CA'].cumsum()

    # Analyse quantitative (80% du CA)
    articles_80 = df_pareto[df_pareto['cumul-part-ca'] <= 80]
    nb_articles_80 = len(articles_80)
    n_articles = len(df_pareto)
    prop_articles_80 = (nb_articles_80 / n_articles) * 100 if n_articles > 0 else 0

    print(f"- Nombre d'articles représentant 80% du CA : {nb_articles_80} / {len(df_pareto)}")
    print(f"- Proportion de ces articles par rapport au catalogue : {prop_articles_80:.2f}%")
       
    # Coordonnées pour la courbe de Lorenz
    x_values = np.arange(1, n_articles + 1) / n_articles * 100 if n_articles > 0 else [0]
    y_values = df_pareto['cumul-part-ca'].values

    # Traçage des courbes
    ax.plot(x_values, y_values, color='#006400', linewidth=2.5, label='Courbe réelle')
    ax.plot([0, 100], [0, 100], color='grey', linestyle='--', label='Égalité parfaite')

    # Point d'inflexion (80/20)
    ax.axvline(x=prop_articles_80, color='red', linestyle=':', alpha=0.6)
    ax.axhline(y=80, color='red', linestyle=':', alpha=0.6)
    ax.scatter(prop_articles_80, 80, color='red', s=80, zorder=5, 
               label=f'{prop_articles_80:.1f}% des articles = 80% CA')

    # Habillage de chaque subplot
    ax.set_title(f'Lorenz : {nom}\n(N = {n_articles} art.)', fontsize=12, fontweight='bold', pad=10)
    ax.set_xlabel('% cumulé des articles', fontsize=10)
    ax.set_ylabel('% cumulé du CA', fontsize=10)
    ax.set_xlim(0, 100)
    ax.set_ylim(0, 100)
    ax.legend(loc='upper left', fontsize=9)
    ax.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()

# Export & Affichage
plt.savefig("Graphiques/CA_Pareto_Comparatif.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# ========================================================================================================
# DOUBLE ANALYSE COMPARATIVE DES OUTLIERS EN PRIX (Global vs Produits Actifs)
# ========================================================================================================

# Définition des deux sous-ensembles
datasets = {
    "df_global": df_final.copy(),
    "df_actifs": df_final[df_final['total_sales'] > 0].copy()
}

resultats = []

for nom, df in datasets.items():
    n = len(df)
    
    # 1. Calculs Z-Score
    moyenne_prix = df['price'].mean()
    ecart_t_prix = df['price'].std()
    seuil_z3 = moyenne_prix + (3 * ecart_t_prix)
    outliers_z = df[df['price'] > seuil_z3]
    nb_z = len(outliers_z)
    pct_z = (nb_z / n) * 100
    
    # 2. Calculs IQR
    q1 = df['price'].quantile(0.25)
    q3 = df['price'].quantile(0.75)
    iqr = q3 - q1
    seuil_max_iqr = q3 + 1.5 * iqr
    outliers_i = df[df['price'] > seuil_max_iqr]
    nb_i = len(outliers_i)
    pct_i = (nb_i / n) * 100
    
    # Stockage des résultats structurés
    resultats.append({
        'Jeu de données': nom,
        'Effectif (N)': n,
        'Seuil Z-Score (>3)': f"{seuil_z3:.2f} €",
        'Nb Outliers (Z-Score)': f"{nb_z} ({pct_z:.2f} %)",
        'Seuil Max IQR': f"{seuil_max_iqr:.2f} €",
        'Nb Outliers (IQR)': f"{nb_i} ({pct_i:.2f} %)"
    })

# Génération automatique du DataFrame de synthèse
synthese_outliers = pd.DataFrame(resultats)
display(synthese_outliers)
```

```python
# ========================================================================================================
# ANALYSE PARETO EN QUANTITÉ (Comparatif Global vs Produits Actifs)
# ========================================================================================================

import pandas as pd

# Définition des deux sous-ensembles
datasets = {
    "Catalogue Global": df_final.copy(),
    "Produits Actifs (total_sales > 0)": df_final[df_final['total_sales'] > 0].copy()
}

synthese_qte = []

for nom, df in datasets.items():
    df_quantites = df.copy()
    
    # Remplacement des NaN par 0 sur les ventes pour éviter les erreurs de calcul
    df_quantites['total_sales'] = df_quantites['total_sales'].fillna(0)
    
    # 1. Tri décroissant par volume de ventes
    df_quantites = df_quantites.sort_values(by='total_sales', ascending=False).reset_index(drop=True)
    
    # 2. Calcul des parts et du cumul en pourcentage (0 à 100%)
    total_qte = df_quantites['total_sales'].sum()
    n_articles = len(df_quantites)
    
    if total_qte > 0 and n_articles > 0:
        df_quantites['Part-quantité (%)'] = (df_quantites['total_sales'] / total_qte) * 100
        df_quantites['Cumul-part-qte (%)'] = df_quantites['Part-quantité (%)'].cumsum()
        
        # 3. Filtrage : produits représentant les premiers 80% du volume
        qte_80 = df_quantites[df_quantites['Cumul-part-qte (%)'] <= 80]
        nbr_produit_80 = len(qte_80)
        volume_ventes_80 = qte_80['total_sales'].sum()
        prop_catalogue = (nbr_produit_80 / n_articles) * 100
    else:
        nbr_produit_80 = 0
        volume_ventes_80 = 0
        prop_catalogue = 0

    # Stockage pour le tableau de synthèse
    synthese_qte.append({
        'Jeu de données': nom,
        'Effectif (N)': n_articles,
        'Volume Total (unités)': f"{total_qte:,.0f}",
        'Nb produits pour 80%': f"{nbr_produit_80} ({prop_catalogue:.2f} %)",
        'Volume 80% cible': f"{volume_ventes_80:,.0f} unités"
    })

# Génération et affichage du tableau comparatif
df_synthese_qte = pd.DataFrame(synthese_qte)
display(df_synthese_qte)
```

```python
# ========================================================================================================
# COURBES DE LORENZ / PARETO EN QUANTITÉ (Comparatif Global vs Produits Actifs)
# ========================================================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Définition des deux sous-ensembles
datasets = {
    "Catalogue Global": df_final.copy(),
    "Produits Actifs (total_sales > 0)": df_final[df_final['total_sales'] > 0].copy()
}

# Création de la figure avec 2 sous-graphiques côte à côte
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

for ax, (nom, df_subset) in zip(axes, datasets.items()):
    df_quantites = df_subset.copy()
    df_quantites['total_sales'] = df_quantites['total_sales'].fillna(0)
    
    # Tri décroissant par volume de ventes
    df_quantites = df_quantites.sort_values(by='total_sales', ascending=False).reset_index(drop=True)
    
    total_qte = df_quantites['total_sales'].sum()
    n_articles = len(df_quantites)
    
    if total_qte > 0 and n_articles > 0:
        df_quantites['Part-quantité (%)'] = (df_quantites['total_sales'] / total_qte) * 100
        df_quantites['Cumul-part-qte (%)'] = df_quantites['Part-quantité (%)'].cumsum()
        
        # Filtrage 80% du volume
        qte_80 = df_quantites[df_quantites['Cumul-part-qte (%)'] <= 80]
        nbr_produit_80 = len(qte_80)
        prop_catalogue = (nbr_produit_80 / n_articles) * 100
    else:
        nbr_produit_80 = 0
        prop_catalogue = 0

    # Coordonnées pour la courbe de Lorenz
    x_values = np.arange(1, n_articles + 1) / n_articles * 100 if n_articles > 0 else [0]
    y_values = df_quantites['Cumul-part-qte (%)'].values if n_articles > 0 else [0]

    # Traçage des courbes
    ax.plot(x_values, y_values, color='#1f77b4', linewidth=2.5, label='Courbe réelle (Quantité)')
    ax.plot([0, 100], [0, 100], color='grey', linestyle='--', label='Égalité parfaite')

    # Point d'inflexion (80% des volumes)
    ax.axvline(x=prop_catalogue, color='red', linestyle=':', alpha=0.6)
    ax.axhline(y=80, color='red', linestyle=':', alpha=0.6)
    ax.scatter(prop_catalogue, 80, color='red', s=80, zorder=5, 
               label=f'{prop_catalogue:.1f}% des art. = 80% des volumes')

    # Habillage
    ax.set_title(f'Lorenz Quantités : {nom}\n(N = {n_articles} art.)', fontsize=12, fontweight='bold', pad=10)
    ax.set_xlabel('% cumulé des articles', fontsize=10)
    ax.set_ylabel('% cumulé des quantités vendues', fontsize=10)
    ax.set_xlim(0, 100)
    ax.set_ylim(0, 100)
    ax.legend(loc='upper left', fontsize=9)
    ax.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()

# Export & Affichage
plt.savefig("Graphiques/Quantites_Pareto_Comparatif.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# ========================================================================================================
# ANALYSE DES STOCKS : ROTATION & VALORISATION FINANCIÈRE (VERSION V2 CORRIGÉE)
# ========================================================================================================

# 1. Calcul de la rotation de stock (en mois)
# On gère les divisions par zéro ou valeurs infinies avec .replace() et .fillna()
df_final['Rotation de stock'] = round(df_final['stock_quantity'] / df_final['total_sales'], 2)
df_final['Rotation de stock'] = df_final['Rotation de stock'].replace([np.inf, -np.inf], np.nan).fillna(0)

# 2. Tri décroissant pour identifier les "Flops" (produits encombrants / stocks dormants)
df_rotation_stock = df_final.sort_values(by='Rotation de stock', ascending=False).reset_index(drop=True)

# 3. Préparation du Flop 20
flop_20 = df_rotation_stock.head(20)

# 4. Création du graphique Matplotlib pour le Flop 20
fig, ax = plt.subplots(figsize=(12, 8))

# Troncature propre des noms de produits
raw_names = flop_20['post_title'].astype(str).values
clean_names = [n.replace('-', ' ')[:38] + '...' if len(n) > 38 else n.replace('-', ' ') for n in raw_names]

# Création des barres horizontales
bars = ax.barh(np.arange(len(clean_names)), flop_20['Rotation de stock'], color='#e74c3c', edgecolor='white', linewidth=0.5)

# Application explicite des noms sur l'axe Y
ax.set_yticks(np.arange(len(clean_names)))
ax.set_yticklabels(clean_names, fontsize=9)

# Ajout des valeurs numériques au bout des barres (avec gestion pour éviter les superpositions)
for bar in bars:
    width = bar.get_width()
    ax.text(width + 0.5, 
            bar.get_y() + bar.get_height()/2, 
            f'{width:,.1f} mois', 
            va='center', ha='left', 
            fontsize=8.5, fontweight='bold', color='#333333')

# Esthétique des axes et du titre
ax.set_title('Flop 20 : Produits avec la plus forte rotation de stock (Stock dormant)', fontsize=14, pad=20, fontweight='bold')

# Nettoyage des bordures
for spine in ['top', 'right']:
    ax.spines[spine].set_visible(False)

# Marge à droite pour les étiquettes et inversion pour avoir le pire en haut
ax.set_xlim(0, flop_20['Rotation de stock'].max() + 5)
ax.invert_yaxis()

# ⚠️ AJUSTEMENT CLÉ : Marge de gauche élargie (35%) pour que les noms longs ne soient jamais coupés
plt.subplots_adjust(left=0.35, right=0.90, top=0.90, bottom=0.1)

# Export du graphique
plt.savefig("Graphiques/rotation_stock.png", dpi=300, bbox_inches="tight")
plt.show()

# ========================================================================================================
# 5. VALORISATION FINANCIÈRE ET VOLUMÉTRIQUE DES STOCKS
# ========================================================================================================

# Valorisation en euros (Quantité restante × Prix d'achat)
df_final['Valorisation_stock_euros'] = df_final['stock_quantity'] * df_final['purchase_price']
valeur_totale_stock = df_final['Valorisation_stock_euros'].sum()

# Volume total de bouteilles en stock
volume_total_stock = df_final['stock_quantity'].sum()

# Affichage propre des indicateurs clés de stock
print(f"📦 Volume total en stock : {volume_total_stock:,.0f} bouteilles")
print(f"💰 Valeur totale du stock immobilisé : {valeur_totale_stock:,.2f} €")
```

```python
# ========================================================================================================
# ÉTAPE 7 - ANALYSE DE LA RENTABILITÉ : TAUX DE MARGE (VERSION TABLEAUX)
# ========================================================================================================


print("=" * 60)
print(" ANALYSE DES TAUX DE MARGE ET DE LA RENTABILITÉ")
print("=" * 60)


# 1. Calculs HT et taux de marge
df_final['prix_ht'] = df_final['price'] / 1.2
df_final['Prix_achat_HT'] = df_final['purchase_price'] / 1.2
df_final['taux_marge'] = ((df_final['prix_ht'] - df_final['Prix_achat_HT']) / df_final['Prix_achat_HT']) * 100

print(f'• Taux de marge minimum global : {df_final["taux_marge"].min():.2f}%')
print(f'• Taux de marge maximum global : {df_final["taux_marge"].max():.2f}%')

# 2. Tableau des anomalies (marge négative)
print('\n--- Produits avec un taux de marge négatif (< 0) ---')
anomalies_marge = df_final.loc[
    df_final['taux_marge'] < 0,
    [
        'product_id',
        'price',
        'stock_quantity',
        'stock_status',
        'purchase_price',
        'taux_marge',
    ],
].copy()

if not anomalies_marge.empty:
  display(anomalies_marge.reset_index(drop=True))
else:
  print('Aucun produit avec une marge négative.')

# 3. Filtrage et tableau agrégé par catégorie
df_marge_positive = df_final.loc[df_final['taux_marge'] > 0].copy()

df_marge_par_type = (
    df_marge_positive.groupby('product_type')['taux_marge']
    .mean()
    .reset_index()
)
df_marge_par_type = df_marge_par_type.sort_values(
    by='taux_marge', ascending=False
).reset_index(drop=True)

#-----------------------------
# Calcul marge positive
#-----------------------------

# 5. Création d'un dataframe filtré avec les taux de marge strictement positifs
df_marge_positive = df_final.loc[(df_final['taux_marge'] > 0), :].copy()

min_marge_positive = df_marge_positive['taux_marge'].min()
max_marge_positive = df_marge_positive['taux_marge'].max()

print(f"\n• Le taux de marge minimum (positif) est de : {min_marge_positive:.4f}%")
print(f"• Le taux de marge maximum (positif) est de : {max_marge_positive:.4f}% \n")


#----------------------------
# Marge par catégorie
#---------------------------

# 6. Agrégation par type de produit (Moyenne du taux de marge)
df_marge_par_type = df_marge_positive.groupby('product_type')['taux_marge'].mean().reset_index()
df_marge_par_type = df_marge_par_type.sort_values(by='taux_marge', ascending=False)

print("\n• Taux de marge moyen par catégorie de produit : ")
print(df_marge_par_type)
print("=" * 60)


#------------------------
# Graphique
#------------------------

# 4. Visualisation graphique
plt.figure(figsize=(10, 6))
sns.barplot(
    x='taux_marge',
    y='product_type',
    data=df_marge_par_type,
    hue='product_type',
    palette='viridis',
    legend=False,
)
plt.title(
    'Taux de marge moyen par catégorie de produit (%)',
    fontsize=14,
    fontweight='bold',
    pad=15,
)
plt.xlabel('Taux de marge moyen (%)', fontsize=11, fontweight='bold')
plt.ylabel('Catégorie', fontsize=11, fontweight='bold')
plt.grid(axis='x', linestyle='--', alpha=0.6)
sns.despine(top=True, right=True)

for index, value in enumerate(df_marge_par_type['taux_marge']):
  plt.text(
      value + 0.5, index, f' {value:.2f}%', va='center', fontsize=9, fontweight='bold'
  )

plt.xlim(0, df_marge_par_type['taux_marge'].max() + 10)
plt.savefig('Graphiques/marge_cat_produit.png', dpi=300, bbox_inches='tight')
plt.show()
```
