```python
#Création d'une colonne "stock_status_2"
#La valeur de cette deuxième colonne sera fonction de la valeur dans la colonne "stock_quantity"
#Si la valeur de la colonne "stock_quantity" est nulle, renseigner "outofstock" sinon mettre "instock"
value = np.where(df_erp['stock_quantity'] > 0, 'instock', 'outofstock')
df_erp.insert(0,"stock_status_2",value)
print(df_erp.columns)
```

```python
#Vérifions que les 2 colonnes sont identiques:
#Les 2 colonnes sont strictement identiques si les valeurs de chaque ligne sont strictement identiques 2 à 2
#Le résultat est l'affichage de True ou False pour chacune des lignes du dataset
#C'est un bon début, mais difficile à exploiter
status = df_erp['stock_status'] == df_erp['stock_status_2']
print(status)

```

```python
#Mais il est possible de synthétiser ce résultat en effectuant la somme de cette colonne:
#True vaut 1 et False 0
#Nous devrions obtenir la somme de 824 qui correspond au nombre de lignes dans ce dataset

# Calcul du nombre de lignes True et False
nb_identiques = (df_erp['stock_status'] == df_erp['stock_status_2']).sum()
nb_different = (df_erp['stock_status'] != df_erp['stock_status_2']).sum()

print(f"Nombre de lignes identiques True : {nb_identiques} / {len(df_erp)}")
print(f"Nombre de lignes différents False: {nb_different} / {len(df_erp)}")
```

```python
#Si les colonnes ne sont absolument pas identiques ligne à ligne alors identifier la ligne en écart

# On filtre là où les deux colonnes sont différentes
df_anomalies = df_erp[df_erp['stock_status'] != df_erp['stock_status_2']]

# Affichage du résultat
print(f"Nombre d'anomalies détectées : {len(df_anomalies)}")
df_anomalies.head()
##Dans ce cas je vous donne ce lien pour apprendre à réaliser des filtres dans Pandas:
##https://bitbucket.org/hrojas/learn-pandas/src/master/
##Lesson 3
```

```python
#Corriger la ou les données incohérentes
#Vérification en utilisant le même code que plus haut pour afficher les problèmes

# Remplacement du statut erroné par le statut calculé
df_erp['stock_status'] = df_erp['stock_status_2']

# Si ce test renvoie un DataFrame vide, votre fichier est parfaitement propre !
anomalies_restantes = df_erp[(df_erp['stock_quantity'] > 0) & (df_erp['stock_status'] == 'outofstock')]
print(f"Nombre d'anomalies restantes : {len(anomalies_restantes)}")
print("Le fichier ERP est désormais cohérent avec les quantités en stock.")
```

<div style="border: 1.5px solid black; padding: 10px;">
    <h4  style="color:Blue;">Définition terme SKU</h4>
    <p>Le terme SKU est l'acronyme de Stock Keeping Unit (en français : Unité de Gestion de Stock).</p>
    <p>C'est un code alphanumérique unique attribué à chaque produit pour permettre de l'identifier précisément dans un inventaire</p>
</div>




```python
# 2) Vérification type de donnée de la colonne SKU
# Fonction pour détecter les différents types de données
def designer_type(valeur):
    # 1. Gestion des valeurs manquantes (NaN ou None)
    if pd.isna(valeur): 
        return 'Manquant (NaN)'
    
    # Conversion en string pour l'analyse de structure
    val_str = str(valeur).strip()
    
    # 2. Identification de la nature du contenu
    if val_str.isdigit(): 
        return 'Numérique'
    if val_str.isalpha(): 
        return 'Alphabétique'
    if val_str.isalnum(): 
        return 'Alphanumérique' # Mélange lettres et chiffres
    
    return 'Spécial (contient symboles/espaces)'

# Application (appel fonction)
type_sku = df_web['sku'].apply(designer_type)
print(type_sku.value_counts())
```

<p> Définition terme SKU : Le terme SKU est l'acronyme de Stock Keeping Unit (en français : Unité de Gestion de Stock). C'est un code alphanumérique unique attribué à chaque produit pour permettre de l'identifier précisément dans un inventaire</p>

<div style="border: 1px solid RGB(51,165,182);" >
<h3 style="margin: auto; padding: 20px; color: RGB(51,165,182); ">Etape 4.2.1 - Identification par le Z-index</h3>
</div>

```python
#Calculer la moyenne du prix
moyenne_prix = df_final['price'].mean()

#Calculer l'écart-type du prix
ecart_t_prix = df_final['price'].std()

#Calculer le Z-score à partir du prix
z_score_prix = (df_final['price'] - moyenne_prix )/ ecart_t_prix

# Aficher les résultats
print(f"- La moyenne du prix = {moyenne_prix:.2f}")
print(f"- L'ecart-type du prix = {ecart_t_prix:.2f}")
print(f"- le Z-score(min) = {z_score_prix.min():.2f} , Z-score(max) = {z_score_prix.max():.2f}")
```

<div style="border: 1.5px solid black; padding: 10px;">
    <h4  style="color:Blue;">L'analyse par le Z-score</h4>
    <p> - Les prix bas restent dans une zone de distribution normale (Z-score min de -1.0) </p>
    <p> - Les prix les plus hauts (Z-score max = 7.23) ne sont pas simplement 'chers', 
            mais constituent des (outliers) au sein du catalogue."</p>
</div>


```python
#Quel est le seuil prix dont le z-score est supérieur à 3?

# Calcul du seuil théorique pour Z = 3 => calcule prix à partir de z-score
seuil_z3 = moyenne_prix + (3 * ecart_t_prix)
print(f"- Le seuil de prix pour un Z-score > 3 est : {seuil_z3:.2f}€")

# Identifier combien de produits dépassent ce seuil
outliers_prestige = df_final.loc[df_final['price'] > seuil_z3,:].copy()

print(f"- Nombre de produits outliers (méthode z-score): {len(outliers_prestige)} sur {len(df_final)} produits \n"
       f"- Proportion des outliers (méthode z-score) : {(len(outliers_prestige) / len(df_final))*100: .2f}%\n" )

# Ajout de la colonne pour voir l'importance de l'écart
outliers_prestige['z_score'] = round((outliers_prestige['price'] - moyenne_prix) / ecart_t_prix , 2)
print(outliers_prestige[['product_id', 'price', 'z_score']].sort_values('z_score', ascending=False))
     
```

<div style="border: 1px solid RGB(51,165,182);" >
<h3 style="margin: auto; padding: 20px; color: RGB(51,165,182); ">Etape 4.2.2 - Identification par l'intervalle interquartile</h3>
</div>

```python
#Utilisation de la fonction "describe" de Pandas pour l'étude des mesures de dispersion
print(df_final['price'].describe())
```

```python
#Définir un seuil pour les articles "outliers" en prix

#1) Méthode IQR
Q1 = df_final["price"].quantile(0.25)
Q3 = df_final["price"].quantile(0.75)
IQR = Q3 -Q1

# calcul des seuils
seuil_min = Q1 - 1.5 * IQR
seuil_max = Q3 + 1.5 * IQR


# Affichage des seuils
print(f"- Q1 = {Q1} \n- Q2 = {Q3} \n- IQR = {IQR} ")
print(f"- Le seuil minimum du prix = {seuil_min: .2f}€")
print(f"- Le seuil maximum du prix = {seuil_max: .2f}€")
```

```python
#Définir le nombre d'articles et la proportion de l'ensemble du catalogue "outliers"

# Détection des outliers prix
df_outliers_price = df_final[df_final["price"] > seuil_max].copy()

#Nombre d'articles
print(f"- Nombre outliers avec la méthode interquartile : {len(df_outliers_price)} sur {len(df_final['price'])} produits")

# Proportion de ce nombre
prop_outliers = len(df_outliers_price) / len(df_final["price"]) * 100
print(f"- Proportion des outiliers avec la méthode interquartile : {prop_outliers: .2f}%")
```

```python
#Selon vous, ces outliers sont-ils justifiés ? Comment le démontrer si cela est possible ?
# Calcul coefficient commercial
df_outliers_price["coef_commercial"] = (
    df_outliers_price["price"] /
    df_outliers_price["purchase_price"]
).round(2)

# Fonction
def commentaire_coef(coef):
    if coef < 1:
        return "vente à perte"
    elif 1 <= coef < 1.5:
        return "marge faible"
    elif 1.5 <= coef < 3:
        return "marge normale"
    elif 3 <= coef < 5:
        return "marge élevée"
    else:  # coef >= 5
        return "prix potentiellement abusif"

# Appliquer la fonction sur la colonne coef_commercial
df_outliers_price["commentaire"] = df_outliers_price["coef_commercial"].apply(commentaire_coef)

# Vérification
print(df_outliers_price[["purchase_price", "price", "coef_commercial", "commentaire"]])
```

<div style="border: 1.5px solid black; padding: 10px;">
<h4  style="color:Blue;">Comparaison des méthodes diagnostic des outliers</h4>
<h5>Méthode z-score basée sur la moyenne</h5>
<p>17 produits dépassant le seuil de 112,28€ sont des valeurs extrêmes et représentent 0,06% du catalogue produits </p>
<h5>Méthode interquartiles basée sur la médiane</h5> 
<p>36 produits dépassant un seil maximum de 83,10€ sont des outliers et représentent 4,35% du catalogue produits</p>
<p> <span  style="font-weight:bold;"> Diagnostic :</span> on remarque que la méthode Z-score est très sensible aux valeurs extrêmes (prix = 225€) ce qui tire la moyenne vers le haut ce qui explique l'ecart d'outliers de 19 produits (36-17)</p>

<h3 style="margin: auto; padding: 20px; color: RGB(51,165,182); ">Etape 5.1 - Analyse des ventes en CA</h3>
</div>

```python
##############################
# Calculer le CA du site web #
##############################

#Créer une colonne calculant le CA par article
df_final['CA'] = df_final['total_sales'] * df_final["price"]

#Calculer la somme de la colonne "ca_par_article"
total_CA = df_final['CA'].sum()

#Ce résultat correspond au chiffre d'affaire du site web
print(f"Le chiffre d'affaires du site web est à hauteur de : {total_CA: .2f}€")
```

<div style="border: 1.5px solid black; padding: 10px;">
<h4  style="color:Blue;">Analyse prix : justification ecart avec le CA (normal) </h4>
<ul>
    <li>CA 'normal' : 143680.1 €</li>
    <li>Mon CA :  143324.10 €</li>
    <li>Soit un écart de 356 €</li>
</ul>
   
<p>Explication:</p>
<p> - Il s'agit de du product_id '5700' du fichier ERP dont la quantité en stock = -1, un statut 'outofstock', et SKU = 14736. De mon coté, je l'ai supprimé en phase nettoyage car on ne peut pas vendre un produit qui est en rupture de stock</p>
<p> - Intérgrer ce produit dans le merge total parce qu'il y'a 8 bouteilles vendues au prix de 44,5€ génère un CA de 356€ expliquant exactement mon écart mais techniquement irréaliste </p>
<p>Décision : Par mesure de prudence et pour garantir la fiabilité l'analyse, j'ai choisi d'exclure cette ligne. Un Chiffre d'Affaires qui ne peut être honoré par une livraison physique est considéré comme une erreur de flux ou une transaction en attente d'annulation.</p>
</div>

```python
#############################
# Calculer le 20 / 80 en CA #
#############################

#Créer une colonne calculant la part du CA de la ligne dans le dataset
df_final['Part CA'] = round((df_final['CA'] / total_CA)*100, 2)

#Tri décroissant de la colonne CA avant de cumuler les valeur
df_final = df_final.sort_values(by='CA', ascending=False)

#Créer une colonne réalisant la somme cumulative de la colonne précedemment créée
df_final['cumul-part-ca'] = round(df_final['Part CA'].cumsum() , 2)

#Grâce aux deux colonnes créées précedemment, calculer le nombre d'articles représentant 80% du CA
# Top articles cumulant 80% du CA
articles_80 = df_final[df_final['cumul-part-ca'] <= 80]

# Nombre d'articles représentant 80% du CA
nb_articles_80 = len(articles_80)

#Afficher la proportion que représente ce groupe d'articles dans le catalogue entier du site web
# Proportion par rapport au catalogue
prop_articles_80 = nb_articles_80 / len(df_final) * 100

print(f"- Nombre d'articles représentant 80% du CA : {nb_articles_80} sur {len(df_final)} articles")
print(f"- Proportion de ces articles par rapport au catalogue : {prop_articles_80:.2f}%")
```

```python
# Graphique CA & Pareto(20/80)
# 1. Préparation des données pour la courbe
# On calcule la proportion cumulée des articles (axe X)
n_articles = len(df_final)
x_values = np.arange(1, n_articles + 1) / n_articles * 100

# On récupère les parts cumulées déjà calculées (axe Y)
y_values = df_final['cumul-part-ca'].values

# 2. Création du graphique
fig, ax = plt.subplots(figsize=(10, 6))

# Tracé de la courbe de Lorenz
ax.plot(x_values, y_values, color='#006400', linewidth=3, label='Courbe de répartition réelle')

# Tracé de la ligne d'équidistribution (le modèle théorique 100% égalitaire)
ax.plot([0, 100], [0, 100], color='grey', linestyle='--', label='Ligne d’égalité parfaite')

# 3. Mise en évidence de TON point 60/80
ax.axvline(x=56.36, color='red', linestyle=':', alpha=0.6)
ax.axhline(y=80, color='red', linestyle=':', alpha=0.6)
ax.scatter(56.36, 80, color='red', s=100, zorder=5, label='Notre point : 52.36% articles = 80% CA')

# 4. Habillage professionnel
ax.set_title('Courbe de Lorenz : Analyse de la concentration du CA', fontsize=14, fontweight='bold', pad=20)
ax.set_xlabel('% cumulé des articles', fontsize=12)
ax.set_ylabel('% cumulé du CA', fontsize=12)
ax.legend(loc='upper left')
ax.grid(True, linestyle='--', alpha=0.5)

# Limites des axes pour la clarté
ax.set_xlim(0, 100)
ax.set_ylim(0, 100)

plt.tight_layout()

#Export graphique
plt.savefig("Graphiques/CA_Pareto.png", dpi=300, bbox_inches="tight")

#Affichage
plt.show()
```


<div style="border: 1.5px solid black; padding: 10px;">    
<h4  style="color:Blue;">Graphique Pareto 20/80 en CA </h4>
<ul>
    <li><span style="font-weight:bold;">La ligne grise (Égalité) :</span> cette ligne représente le modèle d'équidistribution parfaite (égalité), où chaque produit contribuerait de manière égale au chiffre d'affaires (10% des produits font 10% du CA, 50% des produits font 50% du CA, etc). </li>
    <li><span style="font-weight:bold;">La "courbure" douce :</span> dans un business risqué, la courbe formerait un angle droit très serré en haut à gauche. Notre courbe est proche de la ligne grise et montre que chaque référence produit apporte sa valeur au catalogue.</li>
    <li><span style="font-weight:bold;">Le diagnostic :</span> ce graphique prouve qu'on a pas besoin de "modifier" le catalogue, mais de continuer à le valoriser, car la majorité des références sont "utiles" au CA. </li>
</ul>

```python
#TOP 20 des pruduits par prix de vente : pour la présentation
# 1. Configuration de la figure
fig, ax = plt.subplots(figsize=(14, 8))

# 2. Données
ca_values = top20['CA'].values
total_ca = df_final['CA'].sum()
raw_names = top20['post_name'].astype(str).values
clean_names = [n.replace('-', ' ')[:25] + '...' if len(n) > 25 else n.replace('-', ' ') for n in raw_names]
x_pos = np.arange(len(clean_names))

# 3. Couleurs (Viridis)
norm = plt.Normalize(ca_values.min(), ca_values.max())
colors = cm.viridis(norm(ca_values))

# 4. Création des barres
bars = ax.bar(x_pos, ca_values, color=colors, edgecolor='white', linewidth=0.5)

# 5. Placement des informations
for i, bar in enumerate(bars):
    height = bar.get_height()
    perc = (height / total_ca) * 100
    
    # --- INFO 1 : POURCENTAGE EN TÊTE (EN DEHORS) ---
    # On le place légèrement au-dessus de la barre (2% de la hauteur max)
    y_top = height + (ca_values.max() * 0.02)
    ax.text(bar.get_x() + bar.get_width()/2, y_top, 
            f"{perc:.1f}%", 
            ha='center', va='bottom', 
            fontsize=10, fontweight='bold', color='#333333')
    
    # --- INFO 2 : CA AU MILIEU (PLUS PETIT) ---
    # On ajuste la couleur pour la première barre jaune (i=0) pour la visibilité
    text_color_inner = 'black' if i == 0 else 'white'
    
    if height > (ca_values.max() * 0.1): # On ne l'affiche que si la barre est assez grande
        y_mid = height / 2
        ax.text(bar.get_x() + bar.get_width()/2, y_mid, 
                f"{int(height)}", 
                ha='center', va='center', 
                fontsize=8, fontweight='bold', color=text_color_inner,
                alpha=0.9) # Légère transparence pour ne pas surcharger

# 6. Esthétique des axes
ax.set_xticks(x_pos)
ax.set_xticklabels(clean_names, rotation=45, ha='right', fontsize=10)
ax.set_title('Part du CA par produit (Top 20)', fontsize=18, pad=40, fontweight='bold')

# Nettoyage des bordures
for spine in ['top', 'right', 'left']:
    ax.spines[spine].set_visible(False)
ax.get_yaxis().set_visible(False)

# On laisse un peu d'espace en haut pour les étiquettes
ax.set_ylim(0, ca_values.max() * 1.15)

plt.tight_layout()

# Export du graphique
plt.savefig("Graphiques/Top_20_CA_prix_vente.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
#####################################
# Palmarès des articles en quantité #
#####################################

#Effectuer le tri dans l'ordre décroissant de quantités vendues du dataset df_merge
#Réinitialiser l'index du dataset par un reset_index
df_tri_qte = df_final.sort_values('total_sales', ascending=False).reset_index(drop=True)

#Afficher les 20 premiers articles en quantité
top_20_qte = df_tri_qte.head(20)
#print(top_20_qte)
```

```python
#############################
# Calculer le 20 / 80 en CA #
#############################

#Créer une colonne calculant la part en quantité de la ligne dans le dataset
# Calcul de la part de chaque ligne
df_tri_qte['Part-quantité'] = df_tri_qte['total_sales'] / df_tri_qte['total_sales'].sum()

# Cumul (va de 0 à 1.0)
df_tri_qte['Cumul-part_quantité'] = df_tri_qte['Part-quantité'].cumsum()

# Filtrage : on cherche les lignes qui constituent les premiers 80% (0.8)
qte_80 = df_tri_qte[df_tri_qte['Cumul-part_quantité'] <= 0.8]

# RÉSULTATS
nbr_produit_80 = len(qte_80) # Nombre de lignes (produits différents)
volume_ventes_80 = qte_80['total_sales'].sum() # Somme des quantités de ces produits

#Afficher la proportion que représente ce groupe d'articles dans le catalogue entier du site web

print(f"- Nombre de produits nécessaires : {nbr_produit_80} sur {len(df_tri_qte)} produits")
print(f"- Volume des ventes (80% cible) : {volume_ventes_80} unités(quantité)")
print(f"- Proportion de ces articles par rapport au catalogue : {(len(qte_80)/len(df_tri_qte))*100:.2f}%")

```

```python
# Graphique pour la présentation

# 1. On trie les données (pour que le plus vendu soit en haut)
df_plot = top_20_qte.sort_values("total_sales", ascending=True)

# 2. On définit la figure
plt.figure(figsize=(12, 8))

# 3. On trace (barh pour horizontal)
# x = les valeurs (total_sales), y = les étiquettes (post_title)
plt.barh(df_plot["post_title"], df_plot["total_sales"], color= 'rebeccapurple', alpha=0.8)

# 4. Ajout des titres
plt.title("Top 20 des articles les plus vendus (en quantité)", fontsize=14, fontweight='bold')
plt.xlabel("Nombre d'unités vendues")

# 5. Optionnel : ajouter le chiffre exact au bout de chaque barre
for i, v in enumerate(df_plot["total_sales"]):
    plt.text(v + 0.2, i, str(int(v)), va='center', fontweight='bold')

plt.tight_layout()

#Export graphique
plt.savefig("Graphiques/Top20_qte_vendues.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
######################################
# Calculer le nombre de mois de stock #
######################################

#Création de la colonne Rotation de stock
df_final['Rotation de stock'] = round(df_final['stock_quantity'] / df_final['total_sales'],2)

#Remplacement des "inf" par 0
df_final['Rotation de stock'] = df_final['Rotation de stock'].replace([np.inf, -np.inf], 0)

#Effectuer le tri dans l'ordre décroissant du nombre de mois de stock dans le dataset df_merge
# Un produit avec 50 mois de stock est un "Flop" car il encombre l'entrepôt
df_rotation_stock = df_final.sort_values(by='Rotation de stock', ascending=False).reset_index()
#print(df_rotation_stock.head(20))
```

```python
#Graphique en barre du flop 20 des produits qui ont le plus de mois de stock
# 4. Graphique du Flop 20 : les 20 premiers produits avec durée en stock la plus élevée 
flop_20 = df_rotation_stock.head(20)

plt.figure(figsize=(12, 10)) # Un peu plus large pour laisser de la place au texte

# Création des barres
bars = plt.barh(flop_20['post_title'], flop_20['Rotation de stock'], color='salmon')

# Ajout des étiquettes (le nombre de mois) à la fin de chaque barre
for bar in bars:
    width = bar.get_width()
    plt.text(width + 0.5,                         # Position X : un peu après la fin de la barre
             bar.get_y() + bar.get_height()/2,    # Position Y : au milieu de la barre
             f'{width:.1f}',                      # Texte : valeur avec une décimale
             va='center',                         # Alignement vertical
             fontsize=10, 
             fontweight='bold', 
             color='black')

plt.xlabel('Nombre de mois de stock prévus')
plt.title('Flop 20 : Produits avec trop de stock par rapport aux ventes')

# On ajuste les marges pour que le texte ne soit pas coupé à droite
plt.xlim(0, flop_20['Rotation de stock'].max() + 5) 

plt.gca().invert_yaxis() # Garde le pire produit en haut
plt.tight_layout()

#Export graphique
plt.savefig("Graphiques/rotation_stock.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
####################################
# Valorisation des stocks en euros #
####################################

#Création de la colonne Valorisation des stocks en euros
# On multiplie la quantité restante par le prix unitaire
df_final['Valorisation_stock_euros'] = df_final['stock_quantity'] * df_final['purchase_price']

#Calculer la somme de la colonne "Valorisation_stock_euros"
valeur_totale_stock = df_final['Valorisation_stock_euros'].sum()

print(f"- La valeur totale du stock est de : {valeur_totale_stock:,.2f}€")
```

```python
############################
# Analyse du taux de marge #
############################

#Création de la colonne Prix HT
# On suppose une TVA à 20% (diviseur 1.2)
df_final['prix_ht'] = df_final['price'] / 1.2
df_final["Prix_achat_HT"] = df_final["purchase_price"] / 1.2

#Création de la colonne Taux de marge
# Formule : ((Prix de vente HT - Prix d'achat) / Prix d'achat) * 100
df_final['taux_marge'] = (df_final['prix_ht'] - df_final['Prix_achat_HT']) / df_final['Prix_achat_HT']*100

#Afficher le prix minimum de la colonne "taux_marge"
print(f"Le taux de marge minimum est de : {df_final['taux_marge'].min():.4f}%")

#Afficher le prix maximum de la colonne "taux_marge"
print(f"Le taux de marge maximum est de : {df_final['taux_marge'].max():.4f}%")
```

```python
#Affichage de la ligne avec un taux de marge inférieur à 0
print(df_final.loc[(df_final['taux_marge'] < 0) , ['product_id','price','prix_ht','stock_quantity','stock_status','purchase_price','Prix_achat_HT','taux_marge','Valorisation_stock_euros']])
```

```python
#Création d'un dataframe avec les taux positifs
df_marge_positive = df_final.loc[(df_final['taux_marge'] > 0), :].copy()

#Afficher le prix minimum de la colonne "taux_marge"
min_marge_positive = df_marge_positive['taux_marge'].min()

#Afficher le prix maximum de la colonne "taux_marge"
max_marge_positive = df_marge_positive['taux_marge'].max()

print(f"Le taux de marge minimum est de : {min_marge_positive:.4f}% ")
print(f"Le taux de marge maximum est de : {max_marge_positive:.4f}% ")
```

```python
#Création d'un dataframe avec le taux de marge moyen par type de produit
#On groupe par 'product_type' et on calcule la moyenne du taux de marge
df_marge_par_type = df_marge_positive.groupby('product_type')['taux_marge'].mean().reset_index()

# On trie pour faciliter la lecture du graphique
df_marge_par_type = df_marge_par_type.sort_values(by='taux_marge', ascending=False)
#print("Taux de marge moyen par catégorie :")
#print(df_marge_par_type)

#Affichage dans un graphique du taux de marge par type de produit
plt.figure(figsize=(10, 6))

# Correction de la syntaxe selon l'avertissement :
# On ajoute hue='product_type' et legend=False
sns.barplot(
    x='taux_marge', 
    y='product_type', 
    data=df_marge_par_type, 
    hue='product_type', 
    palette='viridis', 
    legend=False
)

plt.title('Taux de marge moyen par catégorie de produit (%)', fontsize=14)
plt.xlabel('Taux de marge moyen (%)')
plt.ylabel('Catégorie produit')
plt.grid(axis='x', linestyle='--', alpha=0.6)

# Ajout des étiquettes de score sur les barres
for index, value in enumerate(df_marge_par_type['taux_marge']):
    plt.text(value, index, f' {value:.2f}', va='center')

#Export graphique
plt.savefig("Graphiques/marge_cat_produit.png", dpi=300, bbox_inches="tight")


plt.show()
```

<h3 style="margin: auto; padding: 20px; color: RGB(51,165,182); ">Etape 5.5 - Analyse des corrélations entre les variables stock, sales et price</h3>
</div>

```python
############################
# Analyse des corrélations #
############################

#Création d'une heatmap de corrélation avec les variables stock, sales et price
#On peut également créer un mask pour n'afficher qu'une demi heatmap

# 1. Sélection des variables clés pour la corrélation
cols_corr = ['stock_quantity','total_sales','price','taux_marge','purchase_price','Rotation de stock']
df_corr = df_marge_positive[cols_corr].corr()

# 2. Création du masque pour masquer la partie supérieure (triangulaire)
mask = np.triu(np.ones_like(df_corr, dtype=bool))

# 3. Configuration de la figure
plt.figure(figsize=(10, 8))

# 4. Création de la heatmap
sns.heatmap(df_corr, 
            mask=mask, 
            annot=True, 
            fmt=".2f", 
            cmap='coolwarm', 
            vmin=-1, vmax=1, 
            center=0,
            square=True, 
            linewidths=.5)

plt.title('Matrice de corrélation : Stocks, Ventes et Prix', fontsize=15)

#Export graphique
plt.savefig("Graphiques/corr_stock_ventes_prix.png", dpi=300, bbox_inches="tight")

plt.show()
```

<div style="border: 1.5px solid black; padding: 10px;">
    <h4  style="color:Blue;">Les corrélations :</h4>    
    <p> <span style= "font-weight:bold;">- Corrélation positive :</span> entre stock_quantity et rotation de stock. Cela confirme que le "Flop 20" est principalement causé par un volume d'unités trop élevé par rapport à la vitesse de vente</p>   
    <p><span style= "font-weight:bold;">- Corrélation négative :</span>  Elle indique que plus le prix (price) est élevé, plus les ventes totales (total_sales) ont tendance à baisser. C'est un argument de poids pour justifier une baisse de prix sur les Champagnes qui dorment en stock.</p>

<p><span style= "font-weight:bold;">- Une corrélation presque parfaite : </span>  entre purchase_price et price. Cela montre que la politique de prix est très rigide : les prix de vente sont fixés directement en fonction des coûts d'achat, sans forcément prendre en compte la demande du marché (ce qui explique parfois les stocks qui stagnent)</p>
